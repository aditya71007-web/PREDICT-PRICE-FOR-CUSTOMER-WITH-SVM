```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score

```

```python
data={'age':[25,30,35,40,45],'monthly_charge':[50,60,70,80,90],'churn':[0,0,1,1,1]}
df=pd.DataFrame(data)
print(df)
```

```python
x=df[['age','monthly_charge']]
y=df['churn']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
svc_model=SVC(kernel='linear',C=1.0)
svc_model.fit(x_train,y_train)
y_pred=svc_model.predict(x_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
print("Classification Report:\n",classification_report(y_test,y_pred))
```

```python
user_age=int(input("Enter your age: "))
user_monthly_charge=float(input("Enter your monthly charge: "))
user_input=np.array([[user_age,user_monthly_charge]])
user_prediction=svc_model.predict(user_input)
if user_prediction[0]==0:
    print("You are not likely to churn.")
else:   print("You are likely to churn.")
```
